## `Natural Language Processing with Deep Learning`

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import string
from collections import Counter
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
text = """
Inference is also costly. Even running a trained model — not training it — requires substantial compute, memory 
bandwidth, and energy. Real-time applications like speech assistants, live translation, or chatbots must respond in 
milliseconds, creating a hard tension between model size (which drives quality) and latency (which drives usability).
Edge deployment is constrained. Running NLP models on mobile phones, embedded devices, or offline systems in areas with 
unreliable internet — again, highly relevant in much of Africa — requires drastically smaller models that inevitably 
sacrifice capability.
Related sub-challenges:
Memory footprint — large models have billions of parameters that must fit in GPU/CPU RAM. A model like GPT-4 cannot 
run on a single consumer device without aggressive compression techniques like quantization or pruning.
"""
```

```python
# Remove punctuation
punc = string.punctuation + "—"

cleaned_txt = [word for word in text if word not in punc]
txt = "".join(cleaned_txt)
```

```python
tokens = word_tokenize(txt)
print(tokens)
```

```python
# Count word frequencies
word_freq = Counter(tokens)
print(word_freq.most_common(10))
```

```python
# Remove stopwords
stop_words = set(stopwords.words('english'))
stopwords_rem = [word for word in tokens if word not in stop_words]
stopwords_rem
```

```python
# Word frequency visual
fig, ax = plt.subplots(figsize=(8, 4))
sns.barplot(x=[word[0] for word in word_freq.most_common(10)], y=[word[1] for word in word_freq.most_common(10)], ax=ax, palette="cool")
ax.set_title("Top 10 Most Common Words")
ax.set_xlabel("Word")
ax.set_ylabel("Frequency")
ax.tick_params(axis="x", rotation=45)
```
