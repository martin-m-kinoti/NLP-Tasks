### `OpenAI Whisper`

```python
import sounddevice as sd
import whisper
import numpy as np
from scipy.io.wavfile import read as wav_read, write

import warnings
warnings.filterwarnings("ignore")
```

```python
duration = 5  # seconds
sample_rate = 16000

print("Recording...")
audio = sd.rec(
    int(duration * sample_rate),
    samplerate=sample_rate,
    channels=1,
    dtype='int16'
)
sd.wait()

write("recording2.wav", sample_rate, audio)

print("Saved recording.wav")
```

```python
def audio_to_text(filename):
    model = whisper.load_model("base")
    sample_rate, audio = wav_read(filename)
    # Whisper expects float32 normalized to [-1, 1] at 16kHz
    audio = audio.astype(np.float32) / 32768.0
    result = model.transcribe(audio)
    return result["text"]
```

```python
# Test1
audio_to_text("recording.wav")
```
