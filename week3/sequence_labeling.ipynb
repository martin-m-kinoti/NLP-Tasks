## `Sequence Labeling`

Sequence labeling is a task where a ML model asigns a label to each word (token) to a sequence of text (statement). Sequence labeling is done through: 
- `Part-of-Speech (POS):` Asigns grammatical tags such as noun, verb, and adjective to each word in a sentence.
- `Named Entity Recognition (NER):` Idntifies and classifies proper nouns in text into predefined categories such as skills, experience, and academic qualifications.
- `Semantic Role Labeling (SRL):` Labels the arguments of a verb in a statement to understand meaning.

### `Part-of-Speech`

```python
import nltk
from nltk.tokenize import word_tokenize
#nltk.download('punkt')
#nltk.download('averaged_perceptron_tagger')
import spacy
import stanza
#stanza.download('en')
```

```python
# POS tagging
statement = "Kinoti's day has been made by the AI panel discussion on current industry trends"
tokens = word_tokenize(statement)
nltk.pos_tag(tokens)
```

### `SpaCy in POS`

`SpaCy` is a tool used for advanced NLP, specially designed for industrial strength production use. It priritizes speed and efficiency, unlike research-focused NLTK

Steps for use:
1. Install the SpaCy library: pip install -U spacy
2. Download a language model: python -m spacy download en_core_web_sm
3. Import and use on text.

```python
# Load the English pipeline
nlp = spacy.load("en_core_web_sm")

# Text processing
text = """
Inference is also costly. Even running a trained model — not training it — requires substantial compute, memory 
bandwidth, and energy. Real-time applications like speech assistants, live translation, or chatbots must respond in 
milliseconds, creating a hard tension between model size (which drives quality) and latency (which drives usability).
Edge deployment is constrained. Running NLP models on mobile phones, embedded devices, or offline systems in areas with 
unreliable internet — again, highly relevant in much of Africa — requires drastically smaller models that inevitably 
sacrifice capability.
Related sub-challenges:
Memory footprint — large models have billions of parameters that must fit in GPU/CPU RAM. A model like GPT-4 cannot 
run on a single consumer device without aggressive compression techniques like quantization or pruning.
"""
doc = nlp(text)
for entity in doc.ents:
    print(entity.text, entity.label_)
```

### `Stanza`

`Stanza` is a collection of accurate and efficient NLP tools developed by the Stanford NLP group. It is designed to take raw text and perform a variety of linguistic analyses across many human languages.

Steps for use:
1. Install the Stanza library: pip install stanza
2. Download an English pipeline: stanza.download('en')
3. Import and use on text.

```python
stanza_nlp = stanza.Pipeline('en', processors='tokenize,ner')
doc = stanza_nlp(text)

for sentence in doc.sentences:
    for entity in sentence.ents:
        print(entity.text, entity.type)
```
