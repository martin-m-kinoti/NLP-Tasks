```python
import torch
from transformers import AutoTokenizer, AutoModelForTokenClassification
from transformers import pipeline

# Load pre-trained BERT model fine-tuned for NER
model_name = "dslim/bert-base-NER"  # BERT fine-tuned on CoNLL-2003
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForTokenClassification.from_pretrained(model_name)

# Create NER pipeline
ner_pipeline = pipeline("ner", model=model, tokenizer=tokenizer)

# Text for sequence labeling
text = "Elon Musk founded SpaceX in California and Tesla is based in Austin"

# Perform sequence labeling
results = ner_pipeline(text)

[result for result in results]
```
