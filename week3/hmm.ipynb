### `Hidden Markov Model`

This is a statistical model used to predict sequences.

NLTK's default POS tagger uses an HMM-based approach

```python
import nltk
from nltk import word_tokenize, pos_tag
import string
```

```python
# HMM POS tagging 
sentence = "Karanja lives in Nairobi and works at Safaricom."
tokens = word_tokenize(sentence)
punct = string.punctuation
clean_tokens = [token for token in tokens if token not in punct]
hmm_output = pos_tag(clean_tokens)

[tag for tag in hmm_output]
```
