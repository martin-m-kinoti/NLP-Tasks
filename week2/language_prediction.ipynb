### `Language Prediction`

```python
import nltk
from collections import Counter, defaultdict
import string
```

```python
# Training text
text = """
Inference is also costly. Even running a trained model — not training it — requires substantial compute, memory 
bandwidth, and energy. Real-time applications like speech assistants, live translation, or chatbots must respond in 
milliseconds, creating a hard tension between model size (which drives quality) and latency (which drives usability).
Edge deployment is constrained. Running NLP models on mobile phones, embedded devices, or offline systems in areas with 
unreliable internet — again, highly relevant in much of Africa — requires drastically smaller models that inevitably 
sacrifice capability.
Related sub-challenges:
Memory footprint — large models have billions of parameters that must fit in GPU/CPU RAM. A model like GPT-4 cannot 
run on a single consumer device without aggressive compression techniques like quantization or pruning.
"""
```

```python
# Tkenize the text
tokens = nltk.word_tokenize(text)

# Remove stopwords
stopwrds = nltk.corpus.stopwords.words('english')
clean_tokens = [token for token in tokens if token not in stopwrds]

# Remove punctuations
punct = string.punctuation + "—"
final_tokens = [token for token in clean_tokens if token not in punct]

# Build bigram model
bigrams = list(nltk.bigrams(final_tokens))
model = defaultdict(Counter)

for word1, word2 in bigrams:
    model[word1][word2] += 1

# Predict the next word
def predict_nxt_word(word):
    if word in model:
        return model[word].most_common(3)
    return "Word not in training data"
```

```python
# Test1
predict_nxt_word("consumer")
```

```python
# Test2
predict_nxt_word("run")
```

```python
# Test3
predict_nxt_word("costly")
```

```python
# Test4
predict_nxt_word("applications")
```

### `Language Prediction using LLM`

```python
from openai import OpenAI
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
api_key = os.getenv("API_KEY")

client = OpenAI(
    api_key=api_key
)

prompt = "Once a upon a time there was"

response = client.responses.create(
    model="gpt-4o",
    input=prompt,
    store=True
)

print(response.output_text)
```
