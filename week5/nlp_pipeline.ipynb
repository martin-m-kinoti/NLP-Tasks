### `Practical NLP Pipeline`

```python
import nltk
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk import pos_tag
```

```python
text = "Natural Language Processing helps computers understand language."

# Tokenization
words = word_tokenize(text)
print("Tokens:", words)

# Remove punctuation
words_clean = [word for word in words if word not in string.punctuation]

# Stopword Removal
stop_words = set(stopwords.words('english'))
filtered = [word for word in words_clean if word.lower() not in stop_words]
print("Filtered:", filtered)

# POS Tagging
pos = pos_tag(filtered)
print("POS Tags:", pos)
```

### `Mini Chatbot Simulation`

```python
print("Welcome to Student Help Center. I am Raven, your assistant. How can I help you?")

while True:
    user = input("You: ")
    if "hello" in user.lower():
        print("Raven: Hello! How can I assist you today?")
    elif "exam" in user.lower():
        print("Raven: You can find exam information and schedules on the university website.")
    elif "bye" in user.lower():
        print("Raven: Goodbye! Have a great day!")
        break
    else:
        print("Raven: I'm sorry, I didn't understand that. Can you please rephrase?")
```
