```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
text = "The lecturer teaches Natural Language Processing"
doc = nlp(text)

for token in doc:
    print(token.text, token.dep_, token.head.text)
```

### `Asignment`

```python
# Dependency analyzer
def dependency_analyzer():
    text = str(input("Enter a sentence: "))
    doc = nlp(text)
    for token in doc:
        print(token.text, token.dep_)
```

```python
# Statement: "The administrator updated student records"
dependency_analyzer()
```

```python
# Statement: "Machine learning improves language processing"
dependency_analyzer()
```

### `OpenAI Depency Analyzer`

```python
from openai import OpenAI
import os
from dotenv import load_dotenv
load_dotenv()

sentence = str(input("Enter a sentence for dependency analysis: "))

client = OpenAI(
    api_key=os.getenv("API_KEY")
)
prompt = f"Perform a dependency analysis on the following sentence: {sentence}"
response = client.responses.create(
    model="gpt-5-nano",
    input=prompt,
    store=True
)

print(response.output_text)
```
