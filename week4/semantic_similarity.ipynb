```python
import spacy
import warnings
warnings.filterwarnings('ignore')
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
def similarity_check():
    statement1 = nlp(str(input("Enter first statement: ")))
    statement2 = nlp(str(input("Enter second statement: ")))

    # Check similarity
    similarity = statement1.similarity(statement2)
    print(f"Similarity score: {similarity}")
```

```python
# Statements 1: "The student passed the examination"
# Statement 2: "The learner succeeded in the exam"
similarity_check()
```

#### `Asignment`

`1. Similar Statements`

```python
# Statements 1: "The student exams were graded by the teacher"
# Statement 2: "The teacher graded the exams"
similarity_check()
```

`2. Unrelated Statements`

```python
# Statements 1: "Trump is the president of the United States"
# Statement 2: "The school is renovating the library"
similarity_check()
```

### `OpenAI in Similarity Analysis`

```python
from openai import OpenAI
import os
from dotenv import load_dotenv
load_dotenv()

client = OpenAI(
    api_key=os.getenv("API_KEY")
)
prompt1 = """
Perform a similarity analysis on the following sentences and only provide a similarity score between 0 and 1, 
where 0 indicates no similarity and 1 indicates identical meaning: 
sentence 1: 'The student exams were graded by the teacher'
sentence 2: 'The teacher graded the exams'
"""
response = client.responses.create(
    model="gpt-5-nano",
    input=prompt1,
    store=True
)

print(response.output_text)
```

```python
prompt2 = """
Perform a similarity analysis on the following sentences and only provide a similarity score between 0 and 1, 
where 0 indicates no similarity and 1 indicates identical meaning:  
sentence 1: 'Trump is the president of the United States'
sentence 2: 'The teacher graded the exams'
"""
response = client.responses.create(
    model="gpt-5-nano",
    input=prompt2,
    store=True
)

print(response.output_text)
```
